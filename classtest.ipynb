```python
plt.figure()
df.iloc[:, 0].value_counts().head(10).plot(kind='bar')
plt.title("Bar Chart")
plt.show()
```

```python
plt.figure()
plt.imshow(df.select_dtypes(include='number').corr())
plt.colorbar()
plt.title("Heat Map")
plt.show()
```

```python
plt.figure()
df.iloc[:, 0].value_counts().head(5).plot(kind='pie', autopct='%1.1f%%')
plt.ylabel("")
plt.title("Pie Chart")
plt.show()
```

```python
plt.figure()
plt.plot(df.select_dtypes(include='number').iloc[:, 0].head(20))
plt.title("Line Chart")
plt.show()
```

```python
num_df = df.select_dtypes(include='number')

plt.figure()
plt.scatter(num_df.iloc[:, 0], num_df.iloc[:, 1])
plt.title("Scatter Plot")
plt.show()
```

```python
plt.figure()
plt.hist(df.select_dtypes(include='number').iloc[:, 0], bins=10)
plt.title("Histogram")
plt.show()
```
