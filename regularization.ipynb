```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.model_selection import train_test_split
from sklearn.metrics import root_mean_squared_error,mean_absolute_error,r2_score,mean_squared_error
from sklearn.datasets import load_diabetes
from sklearn.preprocessing import PolynomialFeatures
```

```python
df = load_diabetes()
X = df.data
Y = df.target
```

```python
print(Y)
```

```python
print(X.shape)
```

```python
lr = LinearRegression()
X_train,X_test,Y_train,Y_test = train_test_split(X,Y,test_size=0.2,random_state=42)
lr.fit(X_train,Y_train)
pred_Y_train = lr.predict(X_train)
pred_Y_test = lr.predict(X_test)

print("MSE (trainning): ",mean_squared_error(Y_train,pred_Y_train))
print("MSE (testing): ",mean_squared_error(Y_test,pred_Y_test))
```

```python
poly = PolynomialFeatures(degree=3)
poly_X = poly.fit_transform(X)
```

```python
lr = LinearRegression()
X_train_p,X_test_p,Y_train,Y_test = train_test_split(X,Y,test_size=0.2,random_state=42)
lr.fit(X_train_p,Y_train)
pred_Y_train_p = lr.predict(X_train_p)
pred_Y_test_p = lr.predict(X_test_p)

print("MSE (trainning): ",mean_squared_error(Y_train,pred_Y_train_p))
print("MSE (testing): ",mean_squared_error(Y_test,pred_Y_test_p))
```

```python
lasso = Lasso(alpha=0.1, max_iter=5000)
X_train_p,X_test_p,Y_train,Y_test = train_test_split(X,Y,test_size=0.2,random_state=42)
lasso.fit(X_train_p,Y_train)
pred_Y_train_p = lasso.predict(X_train_p)
pred_Y_test_p = lasso.predict(X_test_p)

print("MSE (trainning): ",mean_squared_error(Y_train,pred_Y_train_p))
print("MSE (testing): ",mean_squared_error(Y_test,pred_Y_test_p))
```

```python
ridge = Ridge(alpha=0.1, max_iter=5000)
X_train_p,X_test_p,Y_train,Y_test = train_test_split(X,Y,test_size=0.2,random_state=42)
ridge.fit(X_train_p,Y_train)
pred_Y_train_p = ridge.predict(X_train_p)
pred_Y_test_p = ridge.predict(X_test_p)

print("MSE (trainning): ",mean_squared_error(Y_train,pred_Y_train_p))
print("MSE (testing): ",mean_squared_error(Y_test,pred_Y_test_p))
```
