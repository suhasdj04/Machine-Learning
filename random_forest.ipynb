```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.model_selection import train_test_split
from sklearn.metrics import root_mean_squared_error,mean_absolute_error,r2_score,mean_squared_error
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score
from sklearn.ensemble import RandomForestClassifier
import seaborn as sns
```

```python
df = pd.read_csv(r'C:\ML dataset\fruit_classification_dataset.csv')
df.head()
df.info()
df.isnull().sum()
```

```python
cat_cols = ['shape', 'color', 'taste']
for i in cat_cols:
    print(df[i].value_counts())
    
encoded_df = pd.get_dummies(df, columns = cat_cols)
```

```python
encoded_df.head()
```

```python
encoded_df.info()
```

```python
X = encoded_df.drop(columns = 'fruit_name')
Y = encoded_df['fruit_name']

X_train, X_test, Y_train, Y_test = train_test_split(X,Y, test_size=0.2, random_state=42)  #Slitting the data into training and testing sets
rf = RandomForestClassifier(
    max_depth=3,
    n_estimators=50,
    oob_score=False
)

rf.fit(X_train, Y_train)
Y_pred = rf.predict(X_test)
print("Accuracy:", accuracy_score(Y_test, Y_pred))
```

```python
cm = confusion_matrix(Y_test, Y_pred)

sns.heatmap(cm)
plt.show()
```

```python
imp = rf.feature_importances_
index = np.argsort(imp)[::-1]

plt.figure()
plt.bar(range(len(imp)), imp[index])
plt.xticks(range(len(imp)),[encoded_df.columns[i] for i in index], rotation = 45)
plt.show()
```
